```python
from abc import ABC, abstractmethod
class DataReader(ABC):
    @abstractmethod
    def read(self, source):
        pass
class CSVReader(DataReader):
    def read(self, source):
        return f"Reading CSV file: {source}"
class LogReader(DataReader):
    def read(self, source):
        return f"Reading log file: {source}"
class ReaderCreator(ABC):
    @abstractmethod
    def create_reader(self):
        pass
    def process(self, source):
        reader = self.create_reader()
        return reader.read(source)
class CSVReaderCreator(ReaderCreator):
    def create_reader(self):
        return CSVReader()
class LogReaderCreator(ReaderCreator):
    def create_reader(self):
        return LogReader()
creator = CSVReaderCreator()
print(creator.process("sales.csv"))
```

```python
class SmartMartFacade:
    def generate_revenue_report(self, transactions):
        valid_transactions = [
            transaction
            for transaction in transactions
            if transaction["total"] >= 0
        ]
        revenue = sum(
            transaction["total"]
            for transaction in valid_transactions
        )
        return {
            "revenue": revenue,
            "records_processed": len(valid_transactions)
        }
facade = SmartMartFacade()
transactions = [
    {"total": 100},
    {"total": 200},
    {"total": -50}
]
print(facade.generate_revenue_report(transactions))
```

```python
def binary_search(values, target):
    low = 0
    high = len(values) - 1
    while low <= high:
        middle = (low + high) // 2
        if values[middle] == target:
            return middle
        elif values[middle] < target:
            low = middle + 1
        else:
            high = middle - 1
    return -1
numbers = [10, 20, 30, 40, 50]
print(binary_search(numbers, 40))
```

```python
def linear_search(values, target):
    for index, value in enumerate(values):
        if value == target:
            return index
    return -1
numbers = [10, 20, 30, 40, 50]
print(linear_search(numbers, 30))
```

```python
def bubble_sort(values):
    values = values.copy()
    for i in range(len(values)):
        for j in range(len(values) - i - 1):
            if values[j] > values[j + 1]:
                values[j], values[j + 1] = (
                    values[j + 1],
                    values[j]
                )
    return values
print(bubble_sort([30, 10, 40, 20]))
```
